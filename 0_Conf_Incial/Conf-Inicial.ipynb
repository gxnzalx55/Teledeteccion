```python
import getpass
from sentinelhub import  SHConfig
```

Si te equivocas con el identificador, el secret, o el nombre del perfil no hay problema, puedes ejecutar este código tantas veces como sea necesario. También puedes ejecturalo si quieres cambiar el perfil.

Recuerda que te va a pedir por consola el *Client_ID* y la *Client_Secret* que has obtenido en OAuth. 

```python
config = SHConfig()
config.sh_client_id = getpass.getpass('ID:')
config.sh_client_secret = getpass.getpass('Secret:')
config.sh_token_url = "https://identity.dataspace.copernicus.eu/auth/realms/CDSE/protocol/openid-connect/token"
config.sh_base_url = "https://sh.dataspace.copernicus.eu"
config.save("Teledeteccion")
```
