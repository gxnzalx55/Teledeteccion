```python
import cv2
import numpy as np
import matplotlib.pyplot as plt
import time


def histograma(img):
    hist = np.zeros(256, dtype=int)
    filas, cols = img.shape
    for i in range(filas):
        for j in range(cols):
            valor = img[i, j]
            hist[valor] += 1
    return hist


img = cv2.imread("satelite.jpg", cv2.IMREAD_GRAYSCALE)

# 1) Histograma manual
start = time.time()
hist_manual = histograma(img)
end = time.time()
print(f"Tiempo histograma manual: {end - start:.6f} segundos")

# 2) Histograma con NumPy
start = time.time()
hist_numpy, _ = np.histogram(img, bins=256, range=(0,256))
end = time.time()
print(f"Tiempo histograma NumPy: {end - start:.6f} segundos")

# 3) Histograma con OpenCV
start = time.time()
hist_cv = cv2.calcHist([img], [0], None, [256], [0,256])
end = time.time()
print(f"Tiempo histograma OpenCV: {end - start:.6f} segundos")
```
