```python
import cv2
import numpy as np
import matplotlib.pyplot as plt

def histograma(img):
    hist = np.zeros(256, dtype=int)

    filas, cols = img.shape
    for i in range(filas):
        for j in range(cols):
            valor = img[i, j]
            hist[valor] += 1

    plt.figure(figsize=(8,4))
    plt.bar(range(256), hist, width=1.0, color='black')
    plt.title("Histograma de la imagen (escala de grises)")
    plt.xlabel("Intensidad de píxel")
    plt.ylabel("Frecuencia")
    plt.show()

    return hist
```

```python
img = cv2.imread("satelite.jpg", cv2.IMREAD_GRAYSCALE)
hist = histograma(img)
```
