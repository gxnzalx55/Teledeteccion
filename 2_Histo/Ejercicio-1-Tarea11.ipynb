```python
import cv2
import numpy as np
import matplotlib.pyplot as plt

def histograma3B(img_color):
    b, g, r = cv2.split(img_color)

    hist_b = np.zeros(256, dtype=int)
    hist_g = np.zeros(256, dtype=int)
    hist_r = np.zeros(256, dtype=int)

    filas, cols, _ = img_color.shape
    for i in range(filas):
        for j in range(cols):
            hist_b[b[i,j]] += 1
            hist_g[g[i,j]] += 1
            hist_r[r[i,j]] += 1

    plt.figure(figsize=(10,5))
    plt.plot(hist_b, color="blue", label="Canal Azul (B)")
    plt.plot(hist_g, color="green", label="Canal Verde (G)")
    plt.plot(hist_r, color="red", label="Canal Rojo (R)")
    plt.title("Histograma de la imagen RGB")
    plt.xlabel("Nivel de intensidad (0-255)")
    plt.ylabel("Frecuencia")
    plt.legend()
    plt.show()

    return hist_b, hist_g, hist_r

b = cv2.imread("RGB/Landsat_C_02-11_B.png", cv2.IMREAD_GRAYSCALE)
g = cv2.imread("RGB/Landsat_C_03-11_G.png", cv2.IMREAD_GRAYSCALE)
r = cv2.imread("RGB/Landsat_C_04-11_R.png", cv2.IMREAD_GRAYSCALE)

img_rgb = cv2.merge([b, g, r])

plt.imshow(cv2.cvtColor(img_rgb, cv2.COLOR_BGR2RGB))
plt.title("Imagen RGB compuesta")
plt.axis("off")
plt.show()

hist_b, hist_g, hist_r = histograma3B(img_rgb)
```
