```python
#TAREA2 LAMBDA
import math

print(f'sin(pi/4) = {math.sin(math.pi/4)}')

t = [i / 100 for i in range(0, int(round(math.pi, 2) * 100))]
yt = list(map(lambda x: math.sin(x), t))
zt = list(map(lambda x: math.cos(x), t))

print(f't = \n{t}\nyt = [{yt[1]}, ..., {yt[-1]}]\nzt = [{zt[1]}, ..., {zt[-1]}]')
```

```python
#TAREA2 NUMPY (Para cálculos numéricos, tiende a ser mejor NumPy)
import numpy as np
import math

print(f'sin(pi/4) = {math.sin(math.pi/4)}')

t = np.linspace(0, math.pi, 100)
yt = np.sin(t)
zt = np.cos(t)

print(f't = \n{t}\nyt = [{yt[1]}, ..., {yt[-1]}]\nzt = [{zt[1]}, ..., {zt[-1]}]')
```

```python
import matplotlib.pyplot as plt
plt.figure(3)
plt.plot(t,yt)
plt.plot(t,zt,'r--')
plt.xlabel('t (u)')
plt.ylabel('Values (u)')
plt.legend(['sin(t)', 'cos(t)'])
plt.show()
```

```python
#TAREA3
t = np.linspace(-2, 2, 100)
f_t = np.exp(t)      
g_t = np.tan(t)

plt.figure(3)
plt.plot(t, f_t, 'g:')
plt.plot(t, g_t, 'r-.')
plt.xlabel('t (u)')
plt.ylabel('Values (u)')
plt.legend(['f(t) = e^t', 'g(t) = tan(t)'])
plt.ylim(-10, 10)
plt.show()
```
