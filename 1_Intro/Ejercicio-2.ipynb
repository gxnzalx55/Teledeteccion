```python
def factorial1(N):
    f=1
    i=1
    while i<=N:
        f=f*i
        i=i+1
    return f
```

```python
def factorial2(N):
    if N==1:
        return N
    else: return N*factorial2(N-1)
```

```python
#TAREA4
def factorial3(N):
    f = 1
    for i in range(1, N+1):
        f *= i
    return f
```

```python
x = [i for i in range(1, 7)]

print('n\tfactorial1\tfactorial2\tfactorial3')
for n in x:
    print(f'{n}\t{factorial1(n)}\t\t{factorial2(n)}\t\t{factorial3(n)}')
```

```python
import matplotlib.pyplot as plt
import math

def funcion(m, p):
    x, y = [], []
    for aux in range(0, int(m / p)):
        var = aux * p
        x.append(var)
        if 2 <= var <= 3:
            y.append(0)
        else:
            y.append(var)
    plt.figure(1)
    plt.plot(x, y)
    plt.xlabel('x (u)')
    plt.ylabel('y (u)')
    plt.show()

funcion(4, 0.1)
```

```python
import matplotlib.pyplot as plt
import numpy as np

def f(val):
    if 2 <= val <= 3:
        return 0
    else:
        return val
def g(x):
    if x > 0:
        return f(x)
    else:  
        return f(-x)

x_vals = np.arange(-4.0, 4.0, 0.01)
y_vals = [g(val) for val in x_vals]

plt.figure(figsize=(10, 5))
plt.plot(x_vals, y_vals, label='g(x)')
plt.title('Gráfica de la función g(x)')
plt.xlabel('x')
plt.ylabel('g(x)')
plt.grid(True) 
plt.axhline(0, color='black', linewidth=0.5) 
plt.axvline(0, color='black', linewidth=0.5)  
plt.legend()
plt.show()
```
